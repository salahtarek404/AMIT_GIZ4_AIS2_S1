```python
def isPrime(x):
    for i in range (2,x) :
        if x%i==0:
            return False
        return True
    

isPrime(20)    
```

```python
"""
function takes 2 numbers 
get the miniumum
initating a list
loops through the smallest number 
check if there is a reminder for the 2 numbers 
append divisor in a list 
return list
"""

def divisor(num1,num2):
    mn=min(num1,num2)
    commonDivisors=[]
    for i in range (1,mn+1):
        if num1%i==0 and num2%i==0 :
            commonDivisors.append(i)
    return (commonDivisors)

divisor(10,15)
```
