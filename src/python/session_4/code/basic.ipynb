```python
import random

class PasswordGen:
    """
    A class to generate random passwords.
    
    Args:
        length (int): The length of the password to generate.
    """
    def __init__(self, length):
        self.length = length
        self.char = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ1234567890!@#$%^&*"
    
    def passwordNew(self):
        """
        Generate and return a random password.
        
        Returns:
            str: A random password of specified length.
        """
        password = ""
        for _ in range(self.length):
            password += random.choice(self.char)
        return password

if __name__ == "__main__":
    gen = PasswordGen(8)
    generated_password = gen.passwordNew()
    print(generated_password)
```

```python

def counter(words):
    """
    count the number of appearnces for words in a given list
    Args:
    words (list): A list of words
    Returns:
    dict: A dictionary where keys are unique words and values are their counts
    """
    unique_words = set(words)
# print(unique_words)
    counts  = {word:0 for word in unique_words}
    for i in unique_words:
        for j in words:
            if  i == j :
                counts[i] += 1
    return counts
words = ["test","Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]
print(counter(words))
```

```python
import random 

responses={
    "hello":["hi there","welcome"],
    "how are you":["am doing great, hbu?","am alright, U?"],
    "goodbay":["take care","see you","bay"],
    "default":["sorry, i didnt understand","could you please clarify more"]

}
def getResponse(userInput):
    for key in responses:
        if key in userInput:
            return random.choice(responses[key])
    return random.choice(responses["default"])


def chatbot():
    print("hello how are you today?")
    while True:
        userInput=input("User :    ").lower()
        response=getResponse(userInput)
        print(response)

        if userInput=="goodbay":
            break

chatbot()

```
