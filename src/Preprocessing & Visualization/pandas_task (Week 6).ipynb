# Assignment 1:
### - The automated_stat_analyzer Function
- Scenario: A retail company needs a utility to quickly summarize sales data. Students must create a function that identifies the 
"Central Tendency" and "Dispersion" of any numerical column.
- ### Requirements:

* Accept a Pandas DataFrame and a column name.

* Calculate the Mean, Median, and Standard Deviation .

* Identify if the data is "Skewed" by comparing the Mean and Median.


* Bonus: If the column is categorical, return the Mode instead.

### Your Data

```python
import pandas as pd
import numpy as np

# Create a synthetic Company Sales Dataset
data = {
    'Transaction_ID': range(1, 11),
    'Product_Category': ['Electronics', 'Home', 'Electronics', 'Sports', 'Home', 
                         'Electronics', 'Home', 'Sports', 'Electronics', 'Electronics'],
    'Sales_Amount': [150, 200, 155, 300, 210, 180, 205, 1000, 190, 160],
    'Customer_Age': [25, 34, np.nan, 45, 23, 31, 29, np.nan, 38, 40],
    'Rating': [5, 4, 3, 5, 2, 4, 5, 2, 4, 3]
}

df_test = pd.DataFrame(data)
df_test.to_csv('company_sales_test.csv', index=False)

print("\n" + "COMPANY SALES DATASET ".center(60, "="))
print(df_test.to_string())
print("=" * 60 + "\n")


def automated_stat_analyzer(df, column_name):
    """
    Company Task: Provide a summary report of a specific data variable.
    
    Instructions:
    1. Check if the column is numerical or categorical.
    2. For numerical: Calculate Mean, Median, and Standard Deviation.
    3. For categorical: Calculate the Mode.
    4. Return a dictionary with these statistical measures.
    """
    col_data = df[column_name]
    
    if pd.api.types.is_numeric_dtype(col_data):
        mean_val = col_data.mean()
        median_val = col_data.median()
        std_val = col_data.std()
        
        if mean_val > median_val:
            skew = "Right Skewed (Outliers Present)"
        elif mean_val < median_val:
            skew = "Left Skewed"
        else:
            skew = "Symmetric"
            
        return {
            "Mean": round(mean_val, 2),
            "Median": round(median_val, 2),
            "Std Deviation": round(std_val, 2),
            "Skewness": skew
        }
    else:
        return {"Mode": col_data.mode()[0] if len(col_data.mode()) > 0 else "N/A"}


print("\n" + "STATISTICAL ANALYSIS ".center(60, "="))
print("\nNumerical Column (Sales_Amount):")
for key, value in automated_stat_analyzer(df_test, 'Sales_Amount').items():
    print(f"   {key}: {value}")

print("\nCategorical Column (Product_Category):")
for key, value in automated_stat_analyzer(df_test, 'Product_Category').items():
    print(f"   {key}: {value}")
print("=" * 60 + "\n")
```

## Assignment 2: 
  ### The null_handling_strategy Function


#### Scenario: Incoming user data often has missing values.Students must implement a flexible strategy to handle these "Null Values" to prepare data for Machine Learning.
### Requirements:

* Check for null values in the DataFrame.

* Apply a strategy based on parameters: "drop_rows", "fill_mean", or "fill_median" .

* Ensure the function only fills numerical columns when using mean or median.

```python
def null_handling_strategy(df, strategy="fill_mean"):
    """
    Company Task: Clean a dataset by resolving missing (NaN) values.
    
    Strategies:
    - 'drop_rows': Remove rows with missing values
    - 'fill_mean': Fill numerical columns with mean value
    - 'fill_median': Fill numerical columns with median value
    """
    df_clean = df.copy()
    
    if strategy == "drop_rows":
        return df_clean.dropna()
        
    elif strategy == "fill_mean":
        numeric_cols = df_clean.select_dtypes(include=['number']).columns
        df_clean[numeric_cols] = df_clean[numeric_cols].fillna(df_clean[numeric_cols].mean())
        return df_clean
        
    elif strategy == "fill_median":
        numeric_cols = df_clean.select_dtypes(include=['number']).columns
        df_clean[numeric_cols] = df_clean[numeric_cols].fillna(df_clean[numeric_cols].median())
        return df_clean
    
    return df_clean


print("\n" + "NULL HANDLING STRATEGIES ".center(60, "="))
print("\nOriginal Data (with NaN values):")
print(df_test[['Customer_Age', 'Rating']].to_string())

print("\n\nStrategy 1: Fill Mean")
result_mean = null_handling_strategy(df_test, strategy="fill_mean")
print(result_mean[['Customer_Age', 'Rating']].to_string())

print("\n\nStrategy 2: Fill Median")
result_median = null_handling_strategy(df_test, strategy="fill_median")
print(result_median[['Customer_Age', 'Rating']].to_string())

print("\n\nStrategy 3: Drop Rows")
result_drop = null_handling_strategy(df_test, strategy="drop_rows")
print(f"Rows after dropping NaN: {len(result_drop)} (original: {len(df_test)})")
print("=" * 60 + "\n")
```
